```python
import cv2
import matplotlib.pyplot as plt

# Load image
img = cv2.imread("/Users/pr9teen/epics 2/data/train/images/test10_24_jpg.rf.64bfea416b4bc5cc8002a3afa5b17381.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Read label file
with open("/Users/pr9teen/epics 2/data/train/labels/test10_24_jpg.rf.64bfea416b4bc5cc8002a3afa5b17381.txt", "r") as f:
    lines = f.readlines()

# Draw boxes on image
for line in lines:
    class_id, x, y, w, h = map(float, line.split())
    h_img, w_img, _ = img.shape
    x, y, w, h = int(x * w_img), int(y * h_img), int(w * w_img), int(h * h_img)
    
    cv2.rectangle(img, (x, y), (x + w, y + h), (255, 0, 0), 2)  # Blue box

plt.imshow(img)
plt.show()
```
