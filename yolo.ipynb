```python
import os

BASE_DIR = "/Users/pr9teen/epics 2"
os.chdir(BASE_DIR)  # Set working directory
```

```python
from ultralytics import YOLO

# Load the YOLOv8 small model
model = YOLO("yolov8s.pt") 
```

```python
# Train YOLO model for 1 epoch
model.train(data="data.yaml", epochs=1, imgsz=640, plots=True)
```

```python
# Validate the trained model
model.val(data="data.yaml", model="/Users/pr9teen/epics 2/runs/detect/train/weights/best2.pt")
```

```python
# Predict on test images
model.predict(source="/Users/pr9teen/epics 2/data/test/images", conf=0.25, save=True)
```

```python
# Predict on a single test image
model.predict(source="data/testing1.jpg", save=True)
```

```python
# Predict and check if 'Accident' is detected
results = model.predict(source="data/testing1.jpg", conf=0.01)
detected_classes = [model.names[int(box.cls)] for r in results for box in r.boxes]

if "ACCIDENT" in detected_classes:
    print("Accident is present")
else:
    print("No accident detected")
```

```python
import cv2
import matplotlib.pyplot as plt

# Load image
img = cv2.imread("/Users/pr9teen/epics 2/data/train/images/test10_24_jpg.rf.64bfea416b4bc5cc8002a3afa5b17381.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Read label file
with open("/Users/pr9teen/epics 2/data/train/labels/test10_24_jpg.rf.64bfea416b4bc5cc8002a3afa5b17381.txt", "r") as f:
    lines = f.readlines()

# Draw boxes on image
for line in lines:
    class_id, x, y, w, h = map(float, line.split())
    h_img, w_img, _ = img.shape
    x, y, w, h = int(x * w_img), int(y * h_img), int(w * w_img), int(h * h_img)
    
    cv2.rectangle(img, (x, y), (x + w, y + h), (255, 0, 0), 2)  # Blue box

plt.imshow(img)
plt.show()
```
